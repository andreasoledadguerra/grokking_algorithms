## D&C

### D&C algorithms are recurseve algorithms. There are two steps to solving a problem using D&C:

### 1. Figure out the base case. This should be the simplest possible case,
### 2. DivIde or decrease your problem until it becomes the base case. 


##### Euclid's algorithm
##### Tip: when you're writing a recursive function involving an array, the base case is often an empty array or an array with one element.

```python
# Write out the code for the earlier sum function.
def sum(arr):
    total = 0
    for i in arr:
        total += i 
    return total

print(sum([2, 4, 6]))
```

```python
# Write a recursive function to count the number of items in a list.
def count(arr):
    if arr == []:
        return 0
    else:
        return 1 + count(arr[1:])

print(count([1, 2, 3, 4, 5, 6]))
```

```python
# Write a recursive function to find the maximum number in a list.
def maximo(arr):
    if len(arr) == 1: #Base case
        return arr[0]
    resto_max = maximo(arr[1:]) # the rest of the maximum
    if arr[0] > resto_max:
        return arr[0]
    else:
        return resto_max
  

print(maximo([2, 32, 4, 51, 6, 43, 8, 15]))
```

```python
# Base case and recursive case for binary search

def binary_search(arr, item, low, high):
    if low > high: # base case: not found
        return -1
    
    mid = (low + high) // 2
    if arr[mid] == item: # base case: found
        return mid
    elif arr[mid] > item:
        return binary_search(arr, item, low, mid -1) # left
    else:
        return binary_search(arr, item, mid + 1, high) # right

```

## Quicksort

#### Quicksort is a *sorting* algorithm. It's much faster than selection sort, and also uses D&C. Empty array and arrays with just one element don't need to be sorted at all, they will be the base case.
#### If you have an array  of three or more elements, you want to break down this array until you're at the base case. And this is the way to do it: 
#### 1. pick an element from the array : the PIVOT.
#### 2. find the elements smaller than the pivot and the elements larger than the pivot: PARTITIONING.

#### The sub-arrays aren`t sorted. They're just partitioned. But if they were sorted, then sortinng the whole array would be pretty easy:the **quicksort** base case already knows how to sort empty arrays, and it can **recursively** sort arrays of two or more elements. Then combine the results to get a sorted array. This strategy will work with any pivot.

```python
def quicksort(array):
    if len(array) < 2:
        return array # Base case: arrays with 0 or 1 element are already sorted
    else:
        pivot = array[0] # Recursive case
        less = [i for i in array[1:] if i < pivot] #sub-array of all the elements less than the pivot
        greater = [i for i in array [1:] if i > pivot] # sub-array of all the elements greater than the pivot

        return quicksort(less) + [pivot]  + quicksort(greater)

print(quicksort([10, 5, 2, 3]))
```

#### Quicksort is unique because its speed depends on the pivot you choose.

## Merge sort
####  Merge sort is another sorting algorithm which is O(n log n), which is faster than quicksort (O(n2) time in the worst case but in the average case , takes 0(n log n) time).
#### However, a constant can make a difference: often, given th eway quicksort and merge sort are implemented, if they're both 0(n log n) time, quicksort is faster because it hits the average case way more often than the worst case.

#### The best case is also the average case: if you always choose a random element in the array as the pivot, quicksort will complete in 0(n log n) time on average. (There's one exception: if all the elements in your array are the same, you will always hit the worst-case run time without some additional logic.)

#### *Quicksort is one of the fastest sorting algorithms out there, and it's very good example of D&C.
