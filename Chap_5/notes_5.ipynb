## Hash functions

#### A hash function "maps strings to numbers" consistently to the same index, and it knows how big your array is and only returns valid indexes. 

#### Put a hash function and an array together, and you get a data structure called a *hash table*. 

#### * This kind of one-to-one mapping is called an injective function.

#### Python has hash tables; they're called dictionaries.You can make a new hash table using empty braces {} so it has keys and values.

```python
socixs = {}

socixs['Andy'] = 1
socixs['B'] = 1
socixs['Sergio'] = 0

print(socixs['Sergio'])
```

```python
voted = {}

def check_voter(name):
    if name in voted:
        print("Kick them out!")
    else:
        voted[name] = True
        print("Let them vote!")

check_voter("Andy")
check_voter("Eve")
check_voter("Adam")
print(voted)
check_voter("Andy")
```

## Collitions
#### There is *collision* when two keys have been assigned the same slot, and the simplest way to deal with it is multiplying keys map to the same slot, start a linked list at that slot (but if those linked lists get long, it slows down your hash table a lot. But they won't get long if you use a good hash function which give very few collisions).


## Performance
#### A hash table takes constant time, that's means it doesn't matter whether your hash table has 1 element or 1 billion elements-getting something out of a hash table will take the same amount of time. In the average case, hash tables are really fast. In the worst case, a hash table takes O(n)-linear time- for every thing, which is really slow.

#### To avoid collisions, you need:
- A low load factor
- A good hash function

### Load factor
Hash tables use an array for storage, so you count the number of occupied slots in an array. For ex.: 5 arrays and 2 occupied hasa load factor of 2/5.(or 0.4). I f in the best case, each element will get its own slot (i.e.: 100/100), this hash table has a load factor of 1. If we have more elements than slots (i.e.: 50 slots) then it has a load factor of 2. Having a load factor greater than 1 means you have more item than slots in your array. Once the load factor starts to grow, you need to add more slots to your hash table. This is called *resizing* . 

The rule of thumb is to make an array twice the size of the original.
Now you need to reinsert all of those items into this new hash table using the hash function.

Beware, because resizing is expensive but averaged out, hash tables take O(1) even with resizing.

### A good hash function
#### A good hash function distributes values in the array evenly. A bad hash function groupsvalues together and produces a lot of collisions.

```python
# A mapping from battery size to power. The sizes are A, AA, AAA,and AAAA.

sizes = {'A': 4000, 'AA': 2000, 'AAA': 800, 'AAAA': 400}

def battery(size):
    if size not in sizes:
        return "Unknown size"
    power = sizes[size] # base case
    if power >= 4000:
        return "Highest power"
    elif power >= 2000:
        return "High power"
    elif power >= 800:
        return "Medium power"
    else:
        return "Low power"
    

print(battery('AAA'))
```
