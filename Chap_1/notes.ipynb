## Key concepts

- Binary search: algorithm which input is a sorted list of elements. If an element you’re looking for is in that list, this algorithm returns the position where it’s located. Otherwise, binary search returns null. BS needs log n operations to check a list of size n. 

- Logarithms

- Big O notation : special notation that tells you how fast an algorithm is (number of operations -n-); the running time increases as the list size increases. O(log n). 

## Binary Search

```python
def binary_search(arr, item):
    low = 0
    high = len(arr)-1 # low and high keep track of which part of the list you'll search in

    while low <= high: # while you haven't narrowed it down to one element
        mid = (low + high) // 2 # Check the middle element
        guess = arr[mid]
        if guess == item: # Found the item
            return mid
        elif guess > item: # The guess was too high
            high = mid - 1
        else:              # The guess was too low
            low = mid + 1
    return None            # The item doesn't exist


my_list = [1, 3, 5, 7, 9]

print(binary_search(my_list, 3)) # => 1
print(binary_search(my_list, -1)) # => None     
```

## Recap

- Binary search is a lot faster than simple search as your array gets bigger.

- O(log n) is faster than O(n), and it gets a lot faster once the list of items you're searching through grows.

- Algorithm speed isn't measured in seconds.

- Algorithm times are measured in terms of growth of an algorithm.

- Algorithm times are written in big O notation.
