## Key concepts:

- Recursion: is a coding technique where a function calls itself.

- Base case and recursive case. 

- The Stack. The call stack


## Base case and recursive case
#### Every recursive function has two parts: the base case and the recursive case.
#### The recursive case is when the function calls itself. The base case is when the function doesn't call itself again, so it doesn't go into an infinite loop. 



```python
def countdown(i):
    print(i)
    if i <= 1: # <---- Base case
        return i
    else:      # <---- Recursive case
        countdown(i-1)

print(countdown(3))
```

## The stack (simple data structure)

#### When you have a to-do list using, for example, a stack on sticky notes, you read only the topmost item, and it's taken off the list. Your to-do list has only two actions: 

#### -push(insert): add a new item to the top
#### -pop(remove): remove the topmost item and read it.

#### Your computer uses a stack internally called the **call stack**:

```python
def greet2(name):
    print("How are you, " + name + "?")

def bye():
    print("Ok, bye! ")

def greet(name):
    print("Hello, " + name + ";")
    greet2(name)
    print("Getting ready to say bye...")
    bye()

print(greet("Eve"))
```

#### *When you call a function from another function, the calling function is paused in a partially completed state*. All the values of the variables for that function are still stored on the call stack (i.e., in memory).

#### The call stack with recursion

```python
def fact(x):
    if x == 1:
        return 1
    else:
        return x * fact(x-1)
    
print(fact(10))
```

## Recap
